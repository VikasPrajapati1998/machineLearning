## **<font color="red">Compare of Linear & Ridge</font>**

```python
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression, Ridge

X, y = make_regression(n_samples=100, 
                       n_features=1, 
                       n_informative=1, 
                       n_targets=1, 
                       noise=20, 
                       random_state=13)

# ================= Models ==================
print("========== Linear Regression ==========")
lr = LinearRegression()
lr.fit(X, y)
print("w: ", lr.coef_)
print("b: ", lr.intercept_)
print("========== x - x - x ==========\n")

print("========== Ridge Regression (alpha=10) ==========")
rr10 = Ridge(alpha=10)
rr10.fit(X, y)
print("w: ", rr10.coef_)
print("b: ", rr10.intercept_)
print("========== x - x - x ==========\n")

print("========== Ridge Regression (alpha=50) ==========")
rr50 = Ridge(alpha=50)
rr50.fit(X, y)
print("w: ", rr50.coef_)
print("b: ", rr50.intercept_)
print("========== x - x - x ==========\n")

print("========== Ridge Regression (alpha=100) ==========")
rr100 = Ridge(alpha=100)
rr100.fit(X, y)
print("w: ", rr100.coef_)
print("b: ", rr100.intercept_)
print("========== x - x - x ==========\n")


# ==================== Plots ===================
fig, axes = plt.subplots(3, 2, figsize=(10, 10))
fig.suptitle("Linear vs Ridge Regression Models", fontsize=16)

# ---------- Row 1: First Plot ----------
axes[0, 0].scatter(X, y, color="pink")
axes[0, 0].set_title("Scatter Plot (Raw Data)")
axes[0, 0].set_xlabel("X")
axes[0, 0].set_ylabel("y")
axes[0, 0].grid(True)

axes[0, 1].axis("off")

# ---------- Row 2: Second & Third Plots ----------
axes[1, 0].scatter(X, y, color="pink", alpha=0.6, label="Raw Data")  # alpha for opacity/transparency
axes[1, 0].plot(X, lr.predict(X), color="gray", label='Linear (α=0)')
axes[1, 0].legend()
axes[1, 0].grid(True)
axes[1, 0].set_title("Linear Regression")

axes[1, 1].scatter(X, y, color="pink", alpha=0.6, label="Raw Data")  # alpha for opacity/transparency
axes[1, 1].plot(X, rr10.predict(X), color="red", label='Ridge (α=10)')
axes[1, 1].legend()
axes[1, 1].grid(True)
axes[1, 1].set_title("Ridge Regression α=10")

# ---------- Row 3: Fourth & Fifth Plots ----------
axes[2, 0].scatter(X, y, color="pink", alpha=0.6, label="Raw Data")  # alpha for opacity/transparency
axes[2, 0].plot(X, rr50.predict(X), color="green", label='Ridge (α=50)')
axes[2, 0].legend()
axes[2, 0].grid(True)
axes[2, 0].set_title("Ridge Regression α=50")

axes[2, 1].scatter(X, y, color="pink", alpha=0.6, label="Raw Data")  # alpha for opacity/transparency
axes[2, 1].plot(X, rr100.predict(X), color="blue", label='Ridge (α=100)')
axes[2, 1].legend()
axes[2, 1].grid(True)
axes[2, 1].set_title("Ridge Regression α=100")



plt.tight_layout()
plt.subplots_adjust(top=0.92)
plt.show()
```

```python
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression, Ridge

# ================== Dataset ===================
X, y = make_regression(
    n_samples=100,
    n_features=1,
    n_informative=1,
    n_targets=1,
    noise=20,
    random_state=13
)

# ================== Models ====================
models = {
    "Linear (α=0)": LinearRegression(),
    "Ridge (α=10)": Ridge(alpha=10),
    "Ridge (α=50)": Ridge(alpha=50),
    "Ridge (α=100)": Ridge(alpha=100)
}

# Fit models
for name, model in models.items():
    model.fit(X, y)
    print(f"========== {name} ==========")
    print("w:", model.coef_)
    print("b:", model.intercept_)
    print("============================\n")

# ==================== Single Combined Plot ===================
plt.figure(figsize=(8, 6))
plt.scatter(X, y, color='black', alpha=0.6, label='Data Points')

# Plot predictions for each model
plt.plot(X, models["Linear (α=0)"].predict(X), color='red', label='Linear (α=0)')
plt.plot(X, models["Ridge (α=10)"].predict(X), color='orange', label='Ridge (α=10)')
plt.plot(X, models["Ridge (α=50)"].predict(X), color='green', label='Ridge (α=50)')
plt.plot(X, models["Ridge (α=100)"].predict(X), color='blue', label='Ridge (α=100)')

# Titles and labels
plt.title("Linear vs Ridge Regression Comparison", fontsize=14)
plt.xlabel("X")
plt.ylabel("y")
plt.legend(loc="best")
plt.grid(True)
plt.tight_layout()
plt.show()
```

## **<font color="red">Create Custome Ridge Regression</font>**

```python
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
import numpy as np

# Generate synthetic regression data
X, y = make_regression(n_samples=100, 
                       n_features=1, 
                       n_informative=1, 
                       n_targets=1, 
                       noise=20, 
                       random_state=13)

class CustomeRidge:
    def __init__(self, alpha=0.1):
        self.__alpha = alpha
        self.__weight = None
        self.__bias = None
        
    def fit(self, X_train, y_train):
        # Ensure X and y are numpy arrays
        X_train = np.array(X_train).reshape(-1, 1)
        y_train = np.array(y_train)
        
        numerator = 0
        denominator = 0
        for i in range(X_train.shape[0]):
            numerator += ((y_train[i] - y_train.mean()) * (X_train[i] - X_train.mean()))
            denominator += ((X_train[i] - X_train.mean()) * (X_train[i] - X_train.mean()))
        
        denominator += self.__alpha  # Ridge regularization term
        
        self.__weight = numerator / denominator
        self.__bias = y_train.mean() - (self.__weight * X_train.mean())

        print(f"Alpha={self.__alpha}")
        print("Weight:", self.__weight)
        print("Bias:", self.__bias)
        print("====================================")
        
    def predict(self, X_test):
        # Ensure X_test is a numpy array
        X_test = np.array(X_test).reshape(-1, 1)
        # Compute prediction
        return self.__weight * X_test + self.__bias


# ================== Using the custom Ridge ===================
reg0 = CustomeRidge(alpha=0)
reg0.fit(X, y)

reg50 = CustomeRidge(alpha=10)
reg50.fit(X, y)

reg100 = CustomeRidge(alpha=100)
reg100.fit(X, y)

# ================== Predictions ===================
y_pred_0 = reg0.predict(X)
y_pred_50 = reg50.predict(X)
y_pred_100 = reg100.predict(X)

# ================== Plot ===================
plt.scatter(X, y, color='gray', alpha=0.6, label='Data Points')
plt.plot(X, y_pred_0, color='red', label='Custom Ridge α=0')
plt.plot(X, y_pred_50, color='green', label='Custom Ridge α=10')
plt.plot(X, y_pred_100, color='blue', label='Custom Ridge α=100')
plt.xlabel("X")
plt.ylabel("y")
plt.title("Custom Ridge Regression Predictions")
plt.legend()
plt.show()
```
