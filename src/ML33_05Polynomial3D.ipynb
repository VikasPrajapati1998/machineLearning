## **<font color="red">Polynomial Transformation on 3D dataset</font>**
### **<font color="green">3D Dataset</font>**

```python
import numpy as np
import plotly.express as px

# ------------------------------
# 1. Generate 3D data
# ------------------------------
np.random.seed(42)
x = 7 * np.random.rand(100, 1) - 2.8
y = 7 * np.random.rand(100, 1) - 2.8
z = x**2 + y**2 + 0.2*x + 0.2*y + 0.1*x*y + 2 + np.random.randn(100, 1)

# Convert to 1D arrays for plotting
x = x.flatten()
y = y.flatten()
z = z.flatten()

# ------------------------------
# 2. Plot 3D Scatter using Plotly Express
# ------------------------------
fig = px.scatter_3d(
    x=x,
    y=y,
    z=z,
    color=z,                 # Color based on z value
    color_continuous_scale="Viridis",  # Nice gradient color
    title="3D Visualization of Generated Data",
    labels={"x": "X values", "y": "Y values", "z": "Z values"},
    opacity=0.8
)

# Add better layout styling
fig.update_traces(marker=dict(size=6, line=dict(width=0.5, color='DarkSlateGrey')))
fig.update_layout(scene=dict(
    xaxis_title='X-axis',
    yaxis_title='Y-axis',
    zaxis_title='Z-axis',
    aspectmode='cube'
))

# Show the figure
fig.show()
```

### **<font color="blue">Linear Regression Without Polynomial Transformation</font>**

```python
import numpy as np
import plotly.graph_objects as go
from sklearn.linear_model import LinearRegression

# ------------------------------
# 1. Generate 3D data
# ------------------------------
np.random.seed(42)
x = 7 * np.random.rand(100, 1) - 2.8
y = 7 * np.random.rand(100, 1) - 2.8
z = x**2 + y**2 + 0.2*x + 0.2*y + 0.1*x*y + 2 + np.random.randn(100, 1)

# Prepare data for regression
X = np.hstack([x, y])
y_true = z

# ------------------------------
# 2. Linear Regression model
# ------------------------------
model = LinearRegression()
model.fit(X, y_true)
z_pred = model.predict(X)

# Print model parameters
print("Intercept:", model.intercept_[0])
print("Coefficients:", model.coef_[0])

# ------------------------------
# 3. Create regression plane for visualization
# ------------------------------
x_range = np.linspace(x.min(), x.max(), 40)
y_range = np.linspace(y.min(), y.max(), 40)
xx, yy = np.meshgrid(x_range, y_range)
zz = model.intercept_[0] + model.coef_[0][0]*xx + model.coef_[0][1]*yy

# ------------------------------
# 4. Plot original data + regression plane
# ------------------------------
fig = go.Figure()

# Scatter points
fig.add_trace(go.Scatter3d(
    x=x.flatten(),
    y=y.flatten(),
    z=z.flatten(),
    mode='markers',
    marker=dict(size=5, color=z.flatten(), colorscale='Viridis', opacity=0.8),
    name='Data Points'
))

# Regression plane
fig.add_trace(go.Surface(
    x=xx, y=yy, z=zz,
    colorscale='Reds',
    opacity=0.5,
    name='Linear Regression Plane'
))

# Layout
fig.update_layout(
    title="Linear Regression Fit (3D Data)",
    scene=dict(
        xaxis_title='X',
        yaxis_title='Y',
        zaxis_title='Z'
    ),
    width=900,
    height=700
)

fig.show()
```

### **<font color="blue">Linear Regression with Polynomial Transformation</font>**
### **degree = 2**

```python
import numpy as np
import plotly.graph_objects as go
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

# ------------------------------
# 1. Generate 3D data
# ------------------------------
np.random.seed(42)
x = 7 * np.random.rand(100, 1) - 2.8
y = 7 * np.random.rand(100, 1) - 2.8
z = x**2 + y**2 + 0.2*x + 0.2*y + 0.1*x*y + 2 + np.random.randn(100, 1)

# Prepare input data
X = np.hstack([x, y])
y_true = z

# ------------------------------
# 2. Polynomial Transformation (degree = 2)
# ------------------------------
poly = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly.fit_transform(X)

# ------------------------------
# 3. Fit Polynomial Regression Model
# ------------------------------
model = LinearRegression()
model.fit(X_poly, y_true)
z_pred = model.predict(X_poly)

# Print model parameters
print("Intercept:", model.intercept_[0])
print("Coefficients:")
for feature_name, coef in zip(poly.get_feature_names_out(['x', 'y']), model.coef_[0]):
    print(f"  {feature_name:>5} : {coef:.4f}")

# ------------------------------
# 4. Create grid for smooth surface visualization
# ------------------------------
x_range = np.linspace(x.min(), x.max(), 50)
y_range = np.linspace(y.min(), y.max(), 50)
xx, yy = np.meshgrid(x_range, y_range)

# Transform grid into polynomial features
grid = np.c_[xx.ravel(), yy.ravel()]
grid_poly = poly.transform(grid)

# Predict on the grid
zz = model.predict(grid_poly).reshape(xx.shape)

# ------------------------------
# 5. Plot Data + Polynomial Surface
# ------------------------------
fig = go.Figure()

# Scatter data points
fig.add_trace(go.Scatter3d(
    x=x.flatten(),
    y=y.flatten(),
    z=z.flatten(),
    mode='markers',
    marker=dict(size=5, color=z.flatten(), colorscale='Viridis', opacity=0.8),
    name='Data Points'
))

# Polynomial regression surface
fig.add_trace(go.Surface(
    x=xx, y=yy, z=zz,
    colorscale='Reds',
    opacity=0.6,
    name='Polynomial Regression Surface'
))

# Layout
fig.update_layout(
    title="Polynomial Regression (Degree 2) Fit",
    scene=dict(
        xaxis_title='X',
        yaxis_title='Y',
        zaxis_title='Z'
    ),
    width=900,
    height=700
)

fig.show()
```

### **degree = 5**

```python
import numpy as np
import plotly.graph_objects as go
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

# ------------------------------
# 1. Generate 3D data
# ------------------------------
np.random.seed(42)
n_samples = 100
x = 7 * np.random.rand(n_samples, 1) - 2.8
y = 7 * np.random.rand(n_samples, 1) - 2.8
z = x**2 + y**2 + 0.2*x + 0.2*y + 0.1*x*y + 2 + np.random.randn(n_samples, 1)

X = np.hstack([x, y])
y_true = z

# ------------------------------
# 2. Set polynomial degree
# ------------------------------
degree = 5  # change to 3 for degree=3 fit

# ------------------------------
# 3. Fit polynomial regression
# ------------------------------
poly = PolynomialFeatures(degree=degree, include_bias=False)
X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, y_true.ravel())

z_pred = model.predict(X_poly)

# Print model info
print(f"Degree {degree} Polynomial Regression")
print("R^2:", r2_score(y_true, z_pred))
print("MSE:", mean_squared_error(y_true, z_pred))

# ------------------------------
# 4. Create dense grid for smooth surface
# ------------------------------
grid_res = 80
x_lin = np.linspace(x.min(), x.max(), grid_res)
y_lin = np.linspace(y.min(), y.max(), grid_res)
xx, yy = np.meshgrid(x_lin, y_lin)
grid_points = np.c_[xx.ravel(), yy.ravel()]
grid_poly = poly.transform(grid_points)
zz = model.predict(grid_poly).reshape(xx.shape)

# ------------------------------
# 5. Plot 3D scatter + surface
# ------------------------------
fig = go.Figure()

# Scatter points
fig.add_trace(go.Scatter3d(
    x=x.flatten(), y=y.flatten(), z=z.flatten(),
    mode='markers',
    marker=dict(size=5, color=z.flatten(), colorscale='Viridis', opacity=0.8),
    name='Data Points'
))

# Polynomial surface
fig.add_trace(go.Surface(
    x=xx, y=yy, z=zz,
    colorscale='Reds',
    opacity=0.6,
    name=f'Degree {degree} Surface',
    showscale=False
))

fig.update_layout(
    title=f'Polynomial Regression Surface (Degree {degree})',
    scene=dict(xaxis_title='X', yaxis_title='Y', zaxis_title='Z'),
    width=900,
    height=700
)

fig.show()
```
