```python
import pandas as pd

class CustomeLR:
    """
    This is Simple Linear Regression. One Feature One Label.
    It will not work for Multilinear Regression. 
    So do not try for n feature m label.
    """
    def __init__(self):
        self.m = None  # slope/weight
        self.b = None  # intercept/bias

    def fit(self, X_train, y_train):
        if isinstance(X_train, pd.DataFrame):
            X_train = X_train.iloc[:, 0]  # Use first column if DataFrame
        x_mean = X_train.mean()
        y_mean = y_train.mean()

        num = ((X_train - x_mean) * (y_train - y_mean)).sum()
        den = ((X_train - x_mean) ** 2).sum()

        self.m = num / den
        self.b = y_mean - self.m * x_mean
        return self.m, self.b

    def predict(self, X_test):
        if isinstance(X_test, pd.DataFrame):
            X_test = X_test.iloc[:, 0]
        return self.m * X_test + self.b

    def accuracy(self, y_test, y_pred):
        ss_total = ((y_test - y_test.mean()) ** 2).sum()
        ss_residual = ((y_test - y_pred) ** 2).sum()
        r2 = 1 - (ss_residual / ss_total)
        return r2
```

```python
df = pd.read_csv("../data/placement01.csv")
df.head()
```

```python
X = df.iloc[:, 0:1]
y = df.iloc[:, -1]
```

```python
print(type(X))
X.head()
```

```python
print(type(y))
y.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
print(X_train.shape, y_train.shape)
print(X_test.shape, y_test.shape)
```

```python
lr = CustomeLR()
m, b = lr.fit(X_train, y_train)
m, b
```

```python
y_pred = lr.predict(X_test)
y_pred.head()
```

```python
accuracy = lr.accuracy(y_test, y_pred)
accuracy = round(accuracy, 3)
accuracy
```
