## **<font color="blue">Model Predictions and Score</font>**

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix

# data loading
train = pd.read_csv("../data/mnist/mnist_train.csv")
test = pd.read_csv("../data/mnist/mnist_test.csv")

# split X and y correctly
X_train = train.drop(columns=['label'])
y_train = train['label']

X_test = test.drop(columns=['label'])
y_test = test['label']

# model
lrc = LogisticRegression()
dtc = DecisionTreeClassifier()

# train
lrc.fit(X_train, y_train)
dtc.fit(X_train, y_train)

# test
y_pred_lrc = lrc.predict(X_test)
y_pred_dtc = dtc.predict(X_test)

print("="*80)
print("Accuracy Score: ")
print("Logistic_Regression_Accuracy: ", accuracy_score(y_test, y_pred_lrc))
print("Decision_Tree_Accuracy: ", accuracy_score(y_test, y_pred_dtc))
print("="*80)
print("="*80)
print("Confusion Matrix: ")
print("Logistic_Regression_Confusion_Matrix: \n", confusion_matrix(y_test, y_pred_lrc))
print("Decision_Tree_Confusion_Matrix: \n", confusion_matrix(y_test, y_pred_dtc))
print("="*80)
```
