## **<font color="red">Multi-Linear Regression</font>**

```python
import numpy as np
from sklearn.datasets import load_diabetes
```

```python
X, y = load_diabetes(return_X_y=True)
print(X.shape, y.shape)
```

### <font color="aqua">**Using Sklearn's Linear Regression**</font>

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
print(X_train.shape)
print(X_test.shape)
```

```python
reg = LinearRegression()
reg.fit(X_train, y_train)

y_pred = reg.predict(X_test)
r2_score(y_test, y_pred)
```

```python
# get the w(weights) of each columns. Here is 10 columns.
reg.coef_
```

```python
reg.intercept_
```

### <font color="aqua">**Create Custome Multi-Linear Regression Class**</font>

```python
class MLR:
    def __init__(self):
        self.coef_ = None
        self.intercept_ = None

    def fit(self, X_train, y_train):
        X_train = np.insert(X_train, 0, 1, axis=1) # np.insert(matrix, index, value, axis)  # axis=1: column

        # calculate co-efficient
        weights = np.linalg.inv(np.dot(X_train.T, X_train)).dot(X_train.T).dot(y_train)
        self.intercept_ = weights[0]
        self.coef_ = weights[1:]
        
    def predict(self, X_test):
        y_pred = np.dot(X_test, self.coef_) + self.intercept_
        return y_pred
```

```python
lr = MLR()
lr.fit(X_train, y_train)
```

```python
r2_score(y_test, y_pred)
```
